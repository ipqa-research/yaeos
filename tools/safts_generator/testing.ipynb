```python
import feos

import si_units as si

import yaeos

import numpy as np
```

```python
params = feos.Parameters.from_json(["methane", "ethane", "propane", "butane"], "./parameters.json")

params
```

```python
feos_saft = feos.EquationOfState.pcsaft(params)
yaeos_saft = yaeos.PCSAFT(
    m=[1.0, 1.6069, 2.002, 2.3316],
    sigma=[3.7039, 3.5206, 3.6184, 3.7086],
    epsilon_k=[150.03, 191.42, 208.11, 222.88]
)
```

```python
feos_saft
```

```python
n = np.array([5.0, 4.0, 10.0, 15.0])
T = 303.15
V = 0.1 # m^3
```

```python
state = feos.State(
    feos_saft,
    temperature=T*si.KELVIN,
    volume=V*si.METER**3,
    moles=n*si.MOL
)


print("ln_phi:", state.ln_phi())
print("Pressure:", state.pressure())
state.residual_molar_helmholtz_energy_contributions()
```
